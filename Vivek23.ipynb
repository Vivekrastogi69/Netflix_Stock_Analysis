# NETFLIX STOCK ANALYSIS PROJECT

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from datetime import datetime
```

```python
df = pd.read_csv("Netflix.csv")
```

```python
df.head()
```

# volume of Stock Traded

```python
sns.set(rc={'figure.figsize' : (10,5)})
```

# Date wala column ko index mai convert karna hai

```python
df['Date'] =pd.to_datetime(df['Date'])
df = df.set_index('Date')
df.head()
```

```python
sns.lineplot(x=df.index , y = df['Volume'] , label = "volume")
plt.title('Volume of Stock Versus Time')
```

# NETFLIX STOCK PRICE - HIGH , OPEN , CLOSE

```python
df.plot(y = ['High' , 'Close' , 'Open'] , title= "Netflix Stock Price")
```

# NETFLIX STOCKPRICE - DAY , MONTH , YEAR WISEE

```python
fig, (ax1, ax2, ax3) = plt.subplots(3 , figsize = (15,10))
df.groupby(df.index.day).mean().plot(y ="Volume" , ax = ax1 ,xlabel = "Day")
df.groupby(df.index.month).mean().plot(y ="Volume" , ax = ax2 , xlabel = "Month")
df.groupby(df.index.year).mean().plot(y = "Volume" , ax = ax3 , xlabel = "Year")
```

# PLOTTTING THE GRAPH OF HIGHEST AND LOWEST STOCK PRICE

```python
fig,axes = plt.subplots(nrows = 1 , ncols = 2 , sharex = True, figsize = (15,5))
fig.suptitle("High & Low Values Stock Per Period of Time" , fontsize = 20)
sns.lineplot(ax = axes[0] , y = df["High"], x = df.index , color = "green")
sns.lineplot(ax = axes[1] , y = df["High"], x = df.index , color = "red")
```
